```python
# ============================================================
# 🌄 MiDaS Depth Estimation + 3D Point Cloud Visualization (Fixed)
# Works perfectly in Google Colab
# Author: Muhammad Umar (Updated 2025)
# ============================================================

import torch
import cv2
import numpy as np
import open3d as o3d
from torchvision.transforms import Compose, Resize, ToTensor, Normalize
from PIL import Image
import os
from matplotlib import pyplot as plt

# --------------------------------------------
# 1️⃣ Locate Image Automatically
# --------------------------------------------
folder_path = "/content/drive/MyDrive/MAN"

# Find first valid image
valid_exts = ('.jpg', '.jpeg', '.png', '.bmp')
all_images = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.lower().endswith(valid_exts)]

if len(all_images) == 0:
    raise FileNotFoundError("❌ No image found in /content/drive/MyDrive/MAN. Please add .jpg or .png file.")

img_path = all_images[0]
print(f"✅ Using image: {img_path}")

# --------------------------------------------
# 2️⃣ Load MiDaS Model
# --------------------------------------------
midas = torch.hub.load("intel-isl/MiDaS", "DPT_Large")  # Most accurate model
midas_transforms = torch.hub.load("intel-isl/MiDaS", "transforms")

# Use the correct transform
transform = midas_transforms.dpt_transform if hasattr(midas_transforms, "dpt_transform") else midas_transforms.default_transform
midas.eval()

# --------------------------------------------
# 3️⃣ Load and Preprocess Image (Fixed)
# --------------------------------------------
img = cv2.imread(img_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Apply MiDaS transform correctly
input_batch = transform(img).to("cpu")

# --------------------------------------------
# 4️⃣ Run Depth Estimation
# --------------------------------------------
with torch.no_grad():
    prediction = midas(input_batch)
    prediction = torch.nn.functional.interpolate(
        prediction.unsqueeze(1),
        size=img.shape[:2],
        mode="bicubic",
        align_corners=False,
    ).squeeze()

depth = prediction.cpu().numpy()
depth = (depth - depth.min()) / (depth.max() - depth.min())

print("✅ Depth map generated successfully!")

# --------------------------------------------
# 5️⃣ Display Original and Depth Map
# --------------------------------------------
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(depth, cmap="plasma")
plt.title("Estimated Depth Map")
plt.axis("off")
plt.show()

# --------------------------------------------
# 6️⃣ Convert Depth → 3D Point Cloud
# --------------------------------------------
h, w = depth.shape
fx, fy = w / 2, h / 2
cx, cy = w / 2, h / 2

points = []
colors = img / 255.0

for y in range(h):
    for x in range(w):
        z = depth[y, x]
        X = (x - cx) * z / fx
        Y = (y - cy) * z / fy
        points.append([X, -Y, z])

points = np.array(points)
colors = colors.reshape(-1, 3)

# --------------------------------------------
# 7️⃣ Save & Visualize 3D Point Cloud
# --------------------------------------------
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)
pcd.colors = o3d.utility.Vector3dVector(colors)

output_ply = os.path.join(folder_path, "3D_Reconstructed_Image.ply")
o3d.io.write_point_cloud(output_ply, pcd)
print(f"💾 3D point cloud saved: {output_ply}")

# --------------------------------------------
# 8️⃣ Visualize in Colab (Interactive)
# --------------------------------------------
o3d.visualization.draw_geometries([pcd], window_name="3D Depth Visualization")
```

```python
# ============================================================
# 🌄 Enhanced MiDaS 3D Depth Visualization (Colored & Realistic)
# Author: Muhammad Umar (Pro Edition 2025)
# ============================================================

import torch
import cv2
import numpy as np
import open3d as o3d
from torchvision.transforms import Compose, Resize, ToTensor, Normalize
from PIL import Image
import os
from matplotlib import pyplot as plt

# --------------------------------------------
# 1️⃣ Locate Image Automatically
# --------------------------------------------
folder_path = "/content/drive/MyDrive/MAN"

valid_exts = ('.jpg', '.jpeg', '.png', '.bmp')
all_images = [os.path.join(folder_path, f) for f in os.listdir(folder_path) if f.lower().endswith(valid_exts)]

if len(all_images) == 0:
    raise FileNotFoundError("❌ No image found in /content/drive/MyDrive/MAN. Please add .jpg or .png file.")

img_path = all_images[0]
print(f"✅ Using image: {img_path}")

# --------------------------------------------
# 2️⃣ Load MiDaS Model (Most Accurate)
# --------------------------------------------
midas = torch.hub.load("intel-isl/MiDaS", "DPT_Large")
midas_transforms = torch.hub.load("intel-isl/MiDaS", "transforms")
transform = midas_transforms.dpt_transform if hasattr(midas_transforms, "dpt_transform") else midas_transforms.default_transform
midas.eval()

# --------------------------------------------
# 3️⃣ Load Image
# --------------------------------------------
img = cv2.imread(img_path)
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
input_batch = transform(img).to("cpu")

# --------------------------------------------
# 4️⃣ Generate Depth
# --------------------------------------------
with torch.no_grad():
    prediction = midas(input_batch)
    prediction = torch.nn.functional.interpolate(
        prediction.unsqueeze(1),
        size=img.shape[:2],
        mode="bicubic",
        align_corners=False,
    ).squeeze()

depth = prediction.cpu().numpy()
depth = (depth - depth.min()) / (depth.max() - depth.min())

print("✅ Depth map generated successfully!")

# --------------------------------------------
# 5️⃣ Display Results
# --------------------------------------------
plt.figure(figsize=(12,6))
plt.subplot(1,2,1)
plt.imshow(img)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(depth, cmap="plasma")
plt.title("Depth Map (MiDaS)")
plt.axis("off")
plt.show()

# --------------------------------------------
# 6️⃣ Convert Depth → 3D Points
# --------------------------------------------
h, w = depth.shape
fx, fy = w / 2.0, h / 2.0
cx, cy = w / 2.0, h / 2.0

points = []
colors = []

# Downsample for performance
step = 2

for y in range(0, h, step):
    for x in range(0, w, step):
        z = depth[y, x]
        X = (x - cx) * z / fx
        Y = (y - cy) * z / fy
        points.append([X, -Y, z])
        colors.append(img[y, x] / 255.0)

points = np.array(points)
colors = np.array(colors)

# --------------------------------------------
# 7️⃣ Create & Smooth 3D Cloud
# --------------------------------------------
pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(points)
pcd.colors = o3d.utility.Vector3dVector(colors)

# Smoothing (optional for realism)
pcd = pcd.voxel_down_sample(voxel_size=0.002)
pcd.estimate_normals(search_param=o3d.geometry.KDTreeSearchParamHybrid(radius=0.01, max_nn=30))

# --------------------------------------------
# 8️⃣ Add Lighting + Shading
# --------------------------------------------
mesh, densities = o3d.geometry.TriangleMesh.create_from_point_cloud_poisson(pcd, depth=9)
mesh.compute_vertex_normals()

# Light color adjustment (make more 3D visible)
vertex_colors = np.asarray(mesh.vertex_colors)
vertex_colors = np.clip(vertex_colors * 1.2, 0, 1)
mesh.vertex_colors = o3d.utility.Vector3dVector(vertex_colors)

# --------------------------------------------
# 9️⃣ Save + Visualize
# --------------------------------------------
output_ply = os.path.join(folder_path, "Enhanced_3D_Model.ply")
o3d.io.write_triangle_mesh(output_ply, mesh)
print(f"💾 3D model saved: {output_ply}")

# Interactive 3D viewer
o3d.visualization.draw_geometries([mesh], window_name="🧊 Realistic 3D View")
```

```python
# ============================================================
# 🚀 2D Image → Realistic 3D Scene (MiDaS + Open3D + Poisson)
# Single Colab cell. Upload an image when prompted.
# Outputs: Enhanced mesh (.ply, .obj) + rotation video (.mp4)
# ============================================================

# 0) Install & Imports
!pip install -q open3d==0.19.0 imageio[ffmpeg] trimesh pygltflib
import os, sys, math, imageio, numpy as np
from PIL import Image
import matplotlib.pyplot as plt
from google.colab import files
import torch, cv2
import open3d as o3d
import trimesh

# 1) Upload image (or you can set input_path to a file in Drive)
print("📤 Upload your image (jpg/png). Use the upload dialog.")
uploaded = files.upload()
if len(uploaded) == 0:
    raise SystemExit("No file uploaded.")
img_name = list(uploaded.keys())[0]
input_path = "/content/" + img_name
print("Using image:", input_path)

# 2) Load image as RGB numpy
img_pil = Image.open(input_path).convert("RGB")
img_np = np.array(img_pil)
H_orig, W_orig = img_np.shape[:2]
print("Image size:", W_orig, "x", H_orig)

# 3) Load MiDaS (depth estimator)
device = "cuda" if torch.cuda.is_available() else "cpu"
print("Device:", device)
midas = torch.hub.load("intel-isl/MiDaS", "DPT_Large").to(device)
midas_transforms = torch.hub.load("intel-isl/MiDaS", "transforms")
# choose correct transform variant
transform = midas_transforms.dpt_transform if hasattr(midas_transforms, "dpt_transform") else midas_transforms.default_transform

# 4) Prepare input for MiDaS and run inference
img_input = img_np[:, :, ::-1].copy()  # MiDaS code expects BGR? but transforms accept numpy RGB in some versions.
# Use transform on numpy array (works in Colab)
input_tensor = transform(img_input).to(device)
with torch.no_grad():
    prediction = midas(input_tensor)
    prediction = torch.nn.functional.interpolate(
        prediction.unsqueeze(1),
        size=(H_orig, W_orig),
        mode="bicubic",
        align_corners=False,
    ).squeeze()
depth = prediction.cpu().numpy()
# Normalize depth
depth = (depth - depth.min()) / (depth.max() - depth.min() + 1e-8)
print("Depth estimated. min/max:", depth.min(), depth.max())

# 5) Show original + depth (visual)
plt.figure(figsize=(12,5))
plt.subplot(1,2,1); plt.imshow(img_np); plt.title("Original"); plt.axis("off")
plt.subplot(1,2,2); plt.imshow(depth, cmap="magma"); plt.title("MiDaS Depth"); plt.axis("off")
plt.show()

# 6) Build colored point cloud from depth
# We'll downsample pixels for speed (adjust step)
step = 2  # set 1 for full resolution (slower), 2 or 3 speeds up
h, w = depth.shape
fx = 0.8 * w   # focal length estimate (tune if needed)
fy = 0.8 * h
cx = w / 2.0
cy = h / 2.0

pts = []
cols = []
for y in range(0, h, step):
    for x in range(0, w, step):
        z = float(depth[y, x])
        # Convert pixel (x,y) + depth z -> camera coords
        X = (x - cx) * z / fx
        Y = (y - cy) * z / fy
        Z = z
        pts.append([X, -Y, -Z])   # flip axes to look natural
        cols.append(img_np[y, x] / 255.0)

pts = np.asarray(pts, dtype=np.float32)
cols = np.asarray(cols, dtype=np.float32)

pcd = o3d.geometry.PointCloud()
pcd.points = o3d.utility.Vector3dVector(pts)
pcd.colors = o3d.utility.Vector3dVector(cols)

# Optional: remove distant noise via a statistical filter
pcd, ind = pcd.remove_statistical_outlier(nb_neighbors=20, std_ratio=2.0)

# 7) Poisson surface reconstruction (best visually)
# estimate normals first
pcd.estimate_normals(search_param=o3d.geometry.KDTreeSearchParamHybrid(radius=0.02, max_nn=30))
pcd.orient_normals_consistent_tangent_plane(30)

# Poisson meshing
print("Running Poisson surface reconstruction (may take some seconds)...")
mesh, densities = o3d.geometry.TriangleMesh.create_from_point_cloud_poisson(pcd, depth=9)
# Crop mesh by density to remove low-confidence parts
dens = np.asarray(densities)
th = np.percentile(dens, 5)   # tweak percentile to keep main object
verts_to_keep = dens > th
# create mesh mask by removing vertices with low density
vertices = np.asarray(mesh.vertices)
triangles = np.asarray(mesh.triangles)
# remove triangles where any vertex density < th
valid_tri = verts_to_keep[triangles].all(axis=1)
mesh_tri = o3d.geometry.TriangleMesh()
mesh_tri.vertices = o3d.utility.Vector3dVector(vertices)
mesh_tri.triangles = o3d.utility.Vector3iVector(triangles[valid_tri])
# Transfer colors by nearest vertex color from point cloud
# Compute vertex colors by nearest neighbor to point cloud
pcd_tree = o3d.geometry.KDTreeFlann(pcd)
v_colors = []
for v in np.asarray(mesh_tri.vertices):
    _, idx, _ = pcd_tree.search_knn_vector_3d(v, 1)
    v_colors.append(np.asarray(pcd.colors)[idx[0]])
mesh_tri.vertex_colors = o3d.utility.Vector3dVector(np.array(v_colors))

# Smooth mesh a bit for nicer look
mesh_tri = mesh_tri.filter_smooth_taubin(number_of_iterations=5)
mesh_tri.compute_vertex_normals()

# 8) Save outputs (PLY + OBJ)
out_dir = "/content/3d_output"
os.makedirs(out_dir, exist_ok=True)
ply_path = os.path.join(out_dir, "result_mesh.ply")
obj_path = os.path.join(out_dir, "result_mesh.obj")
o3d.io.write_triangle_mesh(ply_path, mesh_tri)
o3d.io.write_triangle_mesh(obj_path, mesh_tri)
print("Saved mesh:", ply_path, obj_path)

# 9) Optional: export glTF via trimesh (for textured glTF you'd need UVs; here we export vertex-colored GLB)
tri_mesh_trimesh = trimesh.Trimesh(vertices=np.asarray(mesh_tri.vertices),
                                   faces=np.asarray(mesh_tri.triangles),
                                   vertex_colors=(np.asarray(mesh_tri.vertex_colors)*255).astype(np.uint8))
glb_path = os.path.join(out_dir, "result_mesh.glb")
tri_mesh_trimesh.export(glb_path)
print("Saved GLB:", glb_path)

# 10) Create offscreen rotation preview (MP4)
W, H = 800, 800
try:
    renderer = o3d.visualization.rendering.OffscreenRenderer(W, H)
    mat = o3d.visualization.rendering.MaterialRecord()
    mat.shader = "defaultLit"
    renderer.scene.set_background([0.95,0.95,0.95,1.0])  # light gray
    renderer.scene.add_geometry("mesh", mesh_tri, mat)
    center = mesh_tri.get_center()
    bbox = mesh_tri.get_max_bound() - mesh_tri.get_min_bound()
    radius = np.linalg.norm(bbox) * 1.2
    frames = []
    for angle in range(0, 360, 8):
        theta = math.radians(angle)
        cam_x = center[0] + radius * math.cos(theta)
        cam_y = center[1] + radius * 0.1   # slight above
        cam_z = center[2] + radius * math.sin(theta)
        renderer.setup_camera(60.0, center, [cam_x, cam_y, cam_z], [0,1,0])
        img_o3d = renderer.render_to_image()
        arr = np.asarray(img_o3d)
        frames.append(arr)
    renderer.release()
    video_path = os.path.join(out_dir, "rotation_preview.mp4")
    imageio.mimwrite(video_path, frames, fps=20)
    print("Rotation preview video saved:", video_path)
except Exception as e:
    print("Offscreen rendering failed (Open3D/driver issue). Error:", e)
    # Fallback: render 2D snapshots using simple matplotlib projections
    video_path = None

# 11) Show results in notebook
print("\n--- Results ---")
print("PLY:", ply_path)
print("OBJ:", obj_path)
print("GLB:", glb_path if os.path.exists(glb_path) else "none")
if video_path and os.path.exists(video_path):
    from IPython.display import Video, display
    display(Video(video_path, embed=True))
else:
    print("No preview video available. Download the mesh and open in Blender/mesh viewer.")

# 12) Offer files for download (colab)
from google.colab import files as colabfiles
colabfiles.download(ply_path)
colabfiles.download(obj_path)
if os.path.exists(glb_path):
    colabfiles.download(glb_path)
if video_path and os.path.exists(video_path):
    colabfiles.download(video_path)

print("✅ Done — downloaded files (if your browser allowed). You can open .obj/.ply/.glb in Blender or Windows 3D Viewer.")
```
